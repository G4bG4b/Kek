```python
from sklearn.linear_model import LinearRegression
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

file_path = '/mnt/data/try.csv' 
df = pd.read_csv('Family Income and Expenditure.csv')


X = df[['House Age']] 
y = df['House Floor Area'] 

model = LinearRegression()

model.fit(X, y)

# Predict house floor area based on house age
y_pred = model.predict(X)


plt.figure(figsize=(12, 8))
plt.scatter(df['House Age'], df['House Floor Area'], color='blue', label='Actual data')
plt.plot(df['House Age'], y_pred, color='red', label='Predicted line')

plt.title('Predicting House Floor Area Based on House Age', fontsize=16)
plt.xlabel('House Age', fontsize=14)
plt.ylabel('House Floor Area', fontsize=14)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

print(f"Model Coefficient: {model.coef_[0]}")
print(f"Model Intercept: {model.intercept_}")  
```

<h>Insight : Older houses are likely to have larger floor areas. This might be because many older houses were built in an era when larger homes were more common, while newer houses could be smaller in comparison due to different design trends, land costs, or smaller family sizes.

House Age appears to have a positive correlation with floor area, though the effect is modest (a small increase in floor area per year of age).</h>
